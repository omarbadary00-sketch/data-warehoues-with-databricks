FUNCTIONS

```python
import pyspark.sql.functions as F
from pyspark.sql.types import StringType
from pyspark.sql.functions import trim , col
```

UPLODE DATA

```python
df = spark.table("workspace.bronze.cust_info")
```

TRIMING DATA

```python
for field in df.schema.fields:
    if isinstance(field.dataType, StringType):
        df = df.withColumn(field.name, trim(col(field.name)))
        
```

CLEARFING DATA

```python
df = (
    df.withColumn(
        "cst_marital_status",
        F.when(F.upper(F.col("cst_marital_status")) == "M", "MARRIED")
         .when(F.upper(F.col("cst_marital_status")) == "D", "DIVORCED")
         .otherwise("N/A")
    )
    .withColumn(
        "cst_gndr",
        F.when(F.upper(F.col("cst_gndr")) == "M", "MALE")
         .when(F.upper(F.col("cst_gndr")) == "F", "FEMALE")
         .otherwise("N/A")
    )
)
```

REMANE CULOMN

```python
RENAME_MAP = {
    "CST_ID":"customer_id",
    "CST_KEY":"customer_key",
    "CST_FIRSTNAME":"customer_firstname",
    "CST_LASTNAME":"customer_lastname",
    "CST_CREATE_DATE":"customer_create_date",
    "CST_GNDR":"customer_gender",
    "CST_MARITAL_STATUS":"customer_marital_status"
}
```

```python
for (old_name, new_name) in RENAME_MAP.items():
    if new_name not in df.columns:
        df = df.withColumnRenamed(old_name, new_name)
```

UPLODE TO SILVER

```python
df 
{
    df.write
      .mode("overwrite")
      .format("delta")
      .saveAsTable("workspace.sliver.cust_info")
}
```
